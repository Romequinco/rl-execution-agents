# SARSA - SUTTON
On-policy TD algorithm: it learns based on its current policy (e.g., e-greedy). Q-learning is an off-policy algorithm because it simply learns the maximimum value (greedy policy), even though this not the way it behaves during training.

On-policy algorithms take a safer path (q-learning opts for the optimal path). These algos take into account the fact there are some random movements in their policy, so they adjust their paths not to fall off the cliff when performing their task.

```python
from collections import defaultdict
from IPython import display
import gym
import matplotlib.pyplot as plt
import numpy as np

from src.utils.plots import plot_training_results

%matplotlib inline
```

```python
class QLearning():

    def __init__(self, env, gamma=0.99, epsilon=0.1, alpha=0.5, 
                 ranges=None, grid_step=20, init_factor = 0.001):
        
        self.env = env
        na = env.action_space.n
        self.Qvalues = defaultdict(lambda: np.random.randn(na) * init_factor) 
        self.alpha = alpha 
        self.epsilon = epsilon 
        self.gamma = gamma 
        self.ranges = ranges
        self.grid_step = grid_step 
        
        
    def encode(self, s):
        if self.ranges is not None:
            to_tuple = np.round(np.divide(s, self.ranges / self.grid_step))
            return tuple(to_tuple.astype(int))
        else:
            return s
            
    def getQs(self, s):
        return self.Qvalues[s] 

    def act(self, s):
        if np.random.rand() > self.epsilon:
            a = self.act_no_explore(s)
        else: 
            a = self.env.action_space.sample()
        return a

    def act_no_explore(self, s):
        return np.argmax(self.getQs(s))
    
    # We simply add an extra input (a1) --> From s, a, r, s1 to --> s, a, r, s1, a1 (sarsa)
    def learn(self, s, a, r, s1, a1, done=False): 
        qsa = self.getQs(s)[a] 
        if done:
            target = r 
        else:
            target = r + self.gamma * self.getQs(s1)[a1] 
           
        self.Qvalues[s][a] = (1 - self.alpha) * qsa + self.alpha * target

        return (target - qsa) ** 2
```

```python
env = gym.make('CartPole-v1')
env.reset()

samples_s = []
for episode in range(500):
    while True:
        a = env.action_space.sample()  
        s1, *_, done = env.step(a)
        samples_s.append(s1)
        if done:
            env.reset()
            break

# grid_range = (np.max(samples_s, axis=0) - np.min(samples_s, axis=0)) # grid 200
grid_range = (np.percentile(samples_s, 90, axis=0) - np.percentile(samples_s, 10, axis=0)) # grid 50

agent = QLearning(env, ranges=grid_range, grid_step=50, gamma=0.995, epsilon=1, alpha=0.1)

nepisodes = 10000
n_log = 1000
min_epsilon = 0.05
epsilon_decay = (1 - min_epsilon) / (nepisodes * 0.9)
history_rewards = []
history_losses = []
for episode in range(nepisodes):
    s = env.reset()
    s = agent.encode(s)
    a = agent.act(s)
    step = 0
    episode_losses = []
    while True:
        s1, r, done, _ = env.step(a)
        s1 = agent.encode(s1)
        a1 = agent.act(s1) # We act again to use this action as an input
        loss = agent.learn(s, a, r, s1, a1, done)
        episode_losses.append(loss)
        step += 1
        a = a1 # a1 is now a for the next step
        s = s1
        if done:
            agent.epsilon = max([agent.epsilon - epsilon_decay, min_epsilon])
            mse = np.mean(episode_losses)
            history_rewards.append(step)
            history_losses.append(mse)
            if not episode % n_log:
                print(
                    f'Episode: {episode}, '
                    f'max_step: {np.round(np.mean(history_rewards[-n_log:]), 2)}, '
                    f'mse: {np.round(np.mean(history_losses[-n_log:]), 2)}, '
                    f'eps: {np.round(agent.epsilon, 2)}'
                )
            break    
```

```python
plot_training_results(
    rewards=history_rewards,
    losses=history_losses,
    window=100,
    scale="log"
)
```
